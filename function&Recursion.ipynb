```python
## Variables length Arguments
## positional  and keywords arguments
## positional argument
def my_function(*args):
    for arg in args:
     print(arg)

print(my_function(1, 2, 4, "four"))
# Output:
# 1
# 2
# 3
# four
```

```python
## keyword arguments
def key_word(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
key_word(name="basit",Class="Bs cs",age="12")     
```

```python
def print_default(*args,**kwargs):
    for val in args:
     print(f"this is a positional argument:{val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")
print_default(1,2,3,4,5,name="basit",age=25,Class="bs cs")        
```

## Example tempreture converter


```python
def temp_coneverter(temp,unit):
  if unit=='C':
   """this function convert temp between C and F"""
   return f"C TO F:{temp * 9/5 +32 }" ## C to F
  elif unit=='F':
    return f" F TO C:{(temp -32)*5/9}" ## F to C
  else:
    return None
temp_coneverter(23,'F')  
```

## Password checker

```python
def password_checker(password):
    if len(password)<5:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char in '!$#%&^*()' for char in password):
        return False
    return True
password_checker("ba3%")
```

## TOTAL COST OF ITHEMS  IN SHOPPING CART

```python
cart=[
         {'name':'apple','price':3,'quantity':4},
         {'name':'banana','price':5,'quantity':7},
         {'name':'orange','price':3,'quantity':6}
         
         
]
def calculate_total_cost(cart):
     totalcost=0
     for item in cart:
          totalcost+=item['price']*item['quantity']
     return totalcost
        

totalcost=calculate_total_cost(cart)
print(f"{totalcost}")
```

## check if  a string is palindrome 
## Original:     a m a n a p l a n a c a n a l p a n a m a
## Cleaned:      amanaplanacanalpanama
##   Reversed:     amanaplanacanalpanama  ← Identical!

## s = "a man a plan a canal panama".lower().replace(" ", "")
## print(s == s[::-1])  # True



```python
def is_palindrome(s):
    s=s.lower().replace(" ","") 
    return  s==s[::-1] 

print(is_palindrome("A man a plan a canal panama"))
print(is_palindrome("Hello"))
```
